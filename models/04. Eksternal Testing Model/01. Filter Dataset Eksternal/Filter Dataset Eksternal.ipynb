```python
import os
import random
import shutil

# ===============================
# PATH DATASET ASLI
# ===============================
source_dir = r"D:\REVA\skripsi\01. Dataset\other dataset\other_dataset_3"

# dataset hasil filter
target_dir = r"D:\REVA\skripsi\01. Dataset\external_dataset_filtered_2"

# ===============================
# MAPPING NAMA KELAS
# ===============================
class_mapping = {
    "2_cataract": "cataract",
    "Glaucoma": "glaucoma",
    "Diabetic Retinopathy": "diabetic_retinopathy",
    "Healthy": "normal"
}

samples_per_class = 100

os.makedirs(target_dir, exist_ok=True)

# ===============================
# FILTER + BALANCE DATASET
# ===============================
for old_class, new_class in class_mapping.items():

    source_class = os.path.join(source_dir, old_class)
    target_class = os.path.join(target_dir, new_class)

    os.makedirs(target_class, exist_ok=True)

    images = os.listdir(source_class)

    print(f"\n{old_class} ditemukan {len(images)} gambar")

    # sampling agar seimbang
    selected_images = random.sample(images, samples_per_class)

    for img in selected_images:

        src = os.path.join(source_class, img)
        dst = os.path.join(target_class, img)

        shutil.copy(src, dst)

    print(f"{new_class} -> {len(selected_images)} gambar disalin")

print("\nDataset eksternal berhasil dibuat!")
```
